```python
# Check the number of records before cleaning
print("Original dataset size:", len(df))

# Check the number of unique records
print("Unique records:", len(df.drop_duplicates()))

# Check class distribution before removing duplicates
print("\nCLASS DISTRIBUTION BEFORE CLEANING:")
print(df["class"].value_counts())

# Create a temporary dataset without exact duplicates
df_unique = df.drop_duplicates()

# Check class distribution after removing duplicates
print("\nCLASS DISTRIBUTION AFTER CLEANING:")
print(df_unique["class"].value_counts())

# Check the new dataset size
print("\nDataset size after removing duplicates:", len(df_unique))
```

```python
# Use the dataset without duplicate records
df_clean = df_unique.copy()

# Reset the row numbers
df_clean = df_clean.reset_index(drop=True)

# Display the cleaned dataset
print("Cleaned dataset shape:", df_clean.shape)

# Display the first five records
print("\nFirst five records:")
display(df_clean.head())

# Check the target distribution
print("\nTarget distribution:")
print(df_clean["class"].value_counts())
```

```python
# Check the unique values in each categorical column

for column in df_clean.columns:
    if df_clean[column].dtype == "object":
        print("\n", column)
        print(df_clean[column].unique())
```

```python
# Separate input features and target

X = df_clean.drop("class", axis=1)
y = df_clean["class"]

print("Input features:")
print(X.head())

print("\nTarget labels:")
print(y.head())

print("\nNumber of input features:", X.shape[1])
print("Number of target records:", y.shape[0])
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the dataset sizes
print("TRAINING DATA:")
print("Input records:", X_train.shape[0])
print("Target records:", y_train.shape[0])

print("\nTESTING DATA:")
print("Input records:", X_test.shape[0])
print("Target records:", y_test.shape[0])

print("\nTotal records:", len(X_train) + len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Identify numerical and categorical columns
numeric_features = ["age"]

categorical_features = [
    column for column in X.columns if column != "age"
]

# Convert categorical values into numerical values
preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Create the machine learning pipeline
model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
model.fit(X_train, y_train)

print("SUCCESS! DiaCare AI model has been trained!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Make predictions using the testing data
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("========== DIACARE AI MODEL EVALUATION ==========")

print("\nModel Accuracy:")
print(f"{accuracy * 100:.2f}%")

# Display classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Display confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Create the Random Forest model
rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        random_state=42
    ))
])

# Train the Random Forest model
rf_model.fit(X_train, y_train)

# Make predictions
rf_pred = rf_model.predict(X_test)

# Calculate accuracy
rf_accuracy = accuracy_score(y_test, rf_pred)

print("========== RANDOM FOREST RESULTS ==========")

print("\nAccuracy:")
print(f"{rf_accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(y_test, rf_pred))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import make_scorer, recall_score

# Create 5-fold cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Select evaluation metrics
scoring = {
    "accuracy": "accuracy",
    "positive_recall": make_scorer(
        recall_score,
        pos_label="Positive"
    ),
    "negative_recall": make_scorer(
        recall_score,
        pos_label="Negative"
    ),
    "f1": make_scorer(
        __import__("sklearn.metrics").metrics.f1_score,
        pos_label="Positive"
    )
}

# Perform cross-validation on training data
cv_results = cross_validate(
    rf_model,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring
)

# Display average results
print("===== 5-FOLD CROSS-VALIDATION =====")

for metric in scoring:
    scores = cv_results["test_" + metric]

    print(
        f"{metric}: "
        f"{scores.mean() * 100:.2f}% "
        f"(± {scores.std() * 100:.2f}%)"
    )
```

```python
from fastapi import FastAPI
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
import pandas as pd
import joblib

# Load the trained AI model
model = joblib.load("diacare_rf_model.pkl")

# Create FastAPI application
app = FastAPI(title="DiaCare AI")

# Allow website communication
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"],
)

# Patient information
class PatientData(BaseModel):
    age: int
    gender: str
    polyuria: str
    polydipsia: str
    sudden_weight_loss: str
    weakness: str
    polyphagia: str
    genital_thrush: str
    visual_blurring: str
    itching: str
    irritability: str
    delayed_healing: str
    partial_paresis: str
    muscle_stiffness: str
    alopecia: str
    obesity: str

# Home page
@app.get("/")
def home():
    return {
        "message": "Welcome to DiaCare AI!",
        "status": "AI backend is running"
    }

# AI screening
@app.post("/predict")
def predict(data: PatientData):

    patient_data = data.model_dump()

    patient_df = pd.DataFrame([patient_data])

    prediction = model.predict(patient_df)[0]

    if prediction == "Positive":
        result = "Higher screening indication"
    else:
        result = "Lower screening indication"

    return {
        "screening_result": result,
        "disclaimer": (
            "This experimental screening result is not a "
            "medical diagnosis. Please consult a qualified "
            "healthcare professional."
        )
    }
```

```python
import gradio as gr
import pandas as pd

def predict_diabetes(
    age, gender, polyuria, polydipsia,
    sudden_weight_loss, weakness, polyphagia,
    genital_thrush, visual_blurring, itching,
    irritability, delayed_healing, partial_paresis,
    muscle_stiffness, alopecia, obesity
):

    patient_data = pd.DataFrame([{
        "age": int(age),
        "gender": gender,
        "polyuria": polyuria,
        "polydipsia": polydipsia,
        "sudden_weight_loss": sudden_weight_loss,
        "weakness": weakness,
        "polyphagia": polyphagia,
        "genital_thrush": genital_thrush,
        "visual_blurring": visual_blurring,
        "itching": itching,
        "irritability": irritability,
        "delayed_healing": delayed_healing,
        "partial_paresis": partial_paresis,
        "muscle_stiffness": muscle_stiffness,
        "alopecia": alopecia,
        "obesity": obesity
    }])

    prediction = model.predict(patient_data)[0]

    if prediction == "Positive":
        result = """
        ## Higher Screening Indication

        Your responses indicate a higher screening indication.

        Please consult a qualified healthcare professional
        for appropriate medical assessment.
        """
    else:
        result = """
        ## Lower Screening Indication

        Your responses indicate a lower screening indication.

        This does not rule out diabetes. Consult a healthcare
        professional if you have concerns or symptoms.
        """

    return result


questions = [
    gr.Number(label="Age", minimum=1, maximum=120, precision=0),
    gr.Dropdown(["Female", "Male"], label="Gender"),
    gr.Dropdown(["Yes", "No"], label="Frequent urination (Polyuria)"),
    gr.Dropdown(["Yes", "No"], label="Excessive thirst (Polydipsia)"),
    gr.Dropdown(["Yes", "No"], label="Sudden weight loss"),
    gr.Dropdown(["Yes", "No"], label="Weakness"),
    gr.Dropdown(["Yes", "No"], label="Excessive hunger (Polyphagia)"),
    gr.Dropdown(["Yes", "No"], label="Genital thrush"),
    gr.Dropdown(["Yes", "No"], label="Blurred vision"),
    gr.Dropdown(["Yes", "No"], label="Itching"),
    gr.Dropdown(["Yes", "No"], label="Irritability"),
    gr.Dropdown(["Yes", "No"], label="Delayed healing"),
    gr.Dropdown(["Yes", "No"], label="Partial paresis"),
    gr.Dropdown(["Yes", "No"], label="Muscle stiffness"),
    gr.Dropdown(["Yes", "No"], label="Hair loss (Alopecia)"),
    gr.Dropdown(["Yes", "No"], label="Obesity")
]

demo = gr.Interface(
    fn=predict_diabetes,
    inputs=questions,
    outputs=gr.Markdown(),
    title="DiaCare AI",
    description="""
    ### AI-Powered Diabetes Risk Screening

    Answer the questions below to receive an experimental
    screening indication.

    **Important:** This tool is for educational screening
    purposes only. It is not a medical diagnosis and does
    not replace professional medical advice.
    """
)

print("DiaCare AI interface is ready!")
```

```python
import re
import pytesseract
from PIL import Image
from pypdf import PdfReader

def scan_report(file):

    if file is None:
        return "Please upload a medical report first."

    filename = file.lower()

    try:
        # Read PDF
        if filename.endswith(".pdf"):

            reader = PdfReader(file)

            extracted_text = ""

            for page in reader.pages:
                extracted_text += page.extract_text() or ""

            if not extracted_text.strip():
                return (
                    "No readable text was found in this PDF. "
                    "It may be a scanned document. "
                    "Please upload a clear JPG or PNG image instead."
                )

        # Read image using OCR
        elif filename.endswith((".png", ".jpg", ".jpeg")):

            image = Image.open(file)

            extracted_text = pytesseract.image_to_string(image)

        else:
            return "Unsupported file. Please upload a PDF, JPG or PNG."

        # Find relevant medical information
        lines = extracted_text.splitlines()

        keywords = ["glucose", "blood sugar", "hba1c", "a1c"]

        results = []

        for line in lines:
            if any(word in line.lower() for word in keywords):
                results.append(line.strip())

        if results:
            findings = "\n".join(
                [f"- {item}" for item in results]
            )
        else:
            findings = "No glucose or HbA1c references were identified."

        return f"""
## 📄 Report Extraction Complete!

### Possible glucose and HbA1c references

{findings}

### Extracted report text

{extracted_text[:4000]}

---

**Important:** This is automated text extraction, not medical interpretation.
Please verify all extracted values, units and reference ranges
against your original report and consult a healthcare professional.
"""

    except Exception as e:
        return f"Unable to read the report. Error: {str(e)}"


print("DiaCare AI report-reading function is ready!")
```

```python
def talk_to_diacare(message, history):

    question = message.lower()

    if any(word in question for word in [
        "chest pain", "difficulty breathing",
        "unconscious", "fainting"
    ]):
        answer = """
        ⚠️ **Please seek emergency medical help immediately.**

        These symptoms may indicate a serious medical problem.
        Do not rely on this chatbot for emergency assistance.
        """

    elif "symptom" in question:
        answer = """
        ### Common diabetes symptoms

        - Frequent urination
        - Excessive thirst
        - Increased hunger
        - Unexplained weight loss
        - Blurred vision
        - Feeling unusually tired

        These symptoms can have other causes too.
        Consult a healthcare professional for assessment.
        """

    elif "hba1c" in question or "a1c" in question:
        answer = """
        ### What is HbA1c?

        HbA1c is a blood test that measures your average
        blood glucose over approximately 2–3 months.

        A healthcare professional can explain your result
        using the appropriate clinical reference ranges.
        """

    elif "glucose" in question or "blood sugar" in question:
        answer = """
        ### What is blood glucose?

        Blood glucose is the amount of sugar present
        in your blood.

        It is an important measurement when screening
        for and monitoring diabetes.

        Test results should be interpreted with their
        units, timing and clinical context.
        """

    elif "prevent" in question:
        answer = """
        ### Healthy habits that may reduce diabetes risk

        - Eat a balanced diet.
        - Stay physically active.
        - Maintain a healthy weight.
        - Get enough sleep.
        - Attend regular health check-ups.

        Healthy habits support overall wellbeing.
        """

    elif "food" in question or "eat" in question:
        answer = """
        ### Healthy eating tips

        - Include vegetables and fruits.
        - Choose whole grains where possible.
        - Include healthy protein sources.
        - Limit sugary drinks.
        - Pay attention to portion sizes.

        A dietitian can help create a suitable meal plan.
        """

    elif "doctor" in question or "consult" in question:
        answer = """
        ### When should you consult a doctor?

        Consider consulting a healthcare professional if
        you experience symptoms such as excessive thirst,
        frequent urination, unexplained weight loss or
        persistent tiredness.

        A professional can recommend appropriate tests.
        """

    else:
        answer = """
        Hello! 👋 I'm DiaCare, your diabetes health-awareness
        assistant.

        I can help you learn about:

        - Diabetes symptoms
        - HbA1c
        - Blood glucose
        - Prevention
        - Healthy eating
        - When to consult a doctor

        Try asking me one of these questions!

        Remember: I provide general health information,
        not medical diagnosis or treatment.
        """

    history = history or []

    history.append({
        "role": "user",
        "content": message
    })

    history.append({
        "role": "assistant",
        "content": answer
    })

    return history, ""

print("DiaCare chatbot is ready!")
```

```python
import pandas as pd
from datetime import datetime

def add_glucose_reading(value, unit, note, history):

    if value is None or value <= 0:
        return "Please enter a valid glucose reading.", history, pd.DataFrame(history)

    history = history or []

    reading = {
        "Date": datetime.now().strftime("%d-%m-%Y %H:%M"),
        "Glucose Reading": value,
        "Unit": unit,
        "Note": note if note else "No note"
    }

    history.append(reading)

    table = pd.DataFrame(history)

    message = "✅ Your glucose reading has been added successfully!"

    return message, history, table

print("DiaCare monitoring function is ready!")
```
